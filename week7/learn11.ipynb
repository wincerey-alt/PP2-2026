```python
## all() 함수는 시퀀스(리스트나 딕셔너리 등)를 받아서, 시퀸스의 모든 항목이 참이면 True를 반환한다

mylist=[1,3,4,6]      #모든 값이 참이다.
all(mylist)
True

mylist=[1,3,4,0]     #하나의 값이 거짓이다.
all(mylist)
False

```

#### all() 함수

```python
## all() 함수는 스퀀스(리스트나 딕셔너리 등)를 받아서,시퀀스의 모든 항목이 참이면 True를 반환한다 

mylist = [1, 3, 4, 6]  # 모든값이참이다. 
all(mylist)
True

mylist = [1, 3, 4, 0]  # 하나의값이거짓이다.
all(mylist)
False

mylist = [True, 0, False, 0]  # 하나의값만참이다. 
all(mylist)
False
```

#### any() 함수

```python
# any() 함수 
# any() 함수는 시퀀스 객체에 있는 한 개의 항목이라도 참인 경우 참을 반환한다.
mylist = [0, 1, 2, 3]                              
any(mylist)
True

mylist = [0, False]  
any(mylist)
False
```

### max(), min() 함수

### enumerate() 함수

```python
# 시퀀스객체를 입력받아, 열거형(enumerate) 객체를 반환한다

seasons = ['Spring', 'Summer', 'Fall', 'Winter']
list(enumerate(seasons))
[(0, 'Spring'), (1, 'Summer'), (2, 'Fall'), (3, 'Winter')]

list(enumerate(seasons, start=1))
[(1, 'Spring'), (2, 'Summer'), (3, 'Fall'), (4, 'Winter')]
filter()

for index, object in enumerate(seasons):
            print(str(index) , “:”  object)
```

### filter() 함수

```python
# filter() 함수는 특정 조건을 만족하는 요소만을 뽑는다.

def myfilter(x):
    returnx > 3

result = filter(myfilter, (1, 2, 3, 4, 5, 6))  
print(list(result)) 

t1 = tuple(range(5: 10))
result2 = filter(lambda x: x > 3, t1)
print(list(result2)) 
```

### Lab: 내장 함수 예제
- 우리는 파티에 적합한 양의 음식을 구입하기 위해서는 총 인원이 몇 명인지 알아야 한다.

```python
invitations = ["Kim", "Lee", "Park", "Choi"]

persons = [1, 3, 0, 6]

sum(persons)
10

# 파티에한사람이라도 오는지를 확인해보자. 이것은any() 함수로 가능하다. 
any(persons)
True
# 이번에는 모든 초대받은 그룹이 전부오는지는 확인해보자. 이것은all() 함수로 가능하다 
all(persons)
False
# 가장많이오는 그룹에는 몇사람이 나있는지를 알아보자. max()로 가능하다. 
max(persons)
6
```

### 정렬과 탐색 
- 파이썬 리스트는 sort()라는 메소드를 가지고 이 메소드는 리스트를 정렬된 상태로 변경한다.

- sorted()라는 내장 함수는 반복 가능한 객체로부터 정렬된 리스트를 새롭게 생성한다.

-  sorted(시퀀스, key함수, reverse여부) : 시퀀스 객체를 받아 원소를 정렬하여 정렬된 시퀀스객체를 생성하고 반환한다. (key함수는각원소의 크기를 반환하는 함수, reverse=True이면 내림차순)

```python
sorted([4, 2, 3, 5, 1]) # 새로운 리스트 생성
[1, 2, 3, 4, 5]

myList = [4, 2, 3, 5, 1]
myList.sort()  # 원래의 리스트 정렬
myList
[1, 2, 3, 4, 5]
```
