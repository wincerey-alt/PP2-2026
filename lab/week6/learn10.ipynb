```python
file_path = "sample.txt"

# 1. 테스트용 샘플 파일 생성
with open(file_path, "w", encoding="utf-8") as f:
    f.write("첫 번째 줄입니다.\n")
    f.write("두 번째 줄입니다.\n")
    f.write("세 번째 줄입니다.\n")

# 2. 파일 포인터를 확인하며 한 줄씩 읽기
with open(file_path, "r", encoding="utf-8") as f:
    line = f.readline().strip()
    
    while line != "":
        print(f"내용: {line}")
        print(f"현재 파일 포인터 위치(byte): {f.tell()}")
        
        # 다음 줄 읽기 (빈 문자열이 반환되면 루프 종료)
        line = f.readline().strip()

    # 파일 끝에 도달한 후, 포인터를 파일 맨 처음(0바이트)으로 되돌리기
    f.seek(0, 0)
    print(f"\nseek(0, 0) 수행 후 포인터 위치: {f.tell()}")
    print(f"다시 읽은 첫 번째 줄: {f.readline().strip()}")
```
