## tkinter란? 

- tkinter는 파이썬에서그래픽사용자인터페이스(GUI: graphical user interface)를 개발할 때 필요한 모듈




### tkinter의 유래 

- tkinter는 예전부터유닉스계열에서사용되던Tcl/Tk 위에객체지향계층을입힌것이다. Tk는John Ousterhout에
의하여Tcl 스크립팅언어를위한GUI 확장으로개발

```python
import tkinter as tk

window=tk.Tk()
window.title("Hello 이승식")
window.geometry("400x200")

label= tk.Label(window,text="Hello tkinter")
label.pack()

window.mainloop() 
```

# 분석 

- from tkinter import *
-> tkinter 모듈을 포함시키는것이다.

- window = Tk()
-> Tk()을호출하면Tk클래스의객체가생성되면서화면에하나의윈도우가생성된다.

- label = Label(window, text="Hello tkinter")
->레이블위젯을생성한다.

- label.pack()
-> pack()은 압축배치관리자를이용하여서레이블을컨테이너에위치시킨다.


### 버튼

```python
import tkinter as tk

root = tk.Tk()
root.title("Oobutton")
root.geometry("600x400")

def on_lbl_clicked(event):
    print(f"라벨이 클릭되었습니다: {event}")

label = tk.Label(root, text="객체지향윈도우프로그래밍")
label.bind("<Double-1>", on_lbl_clicked)
label.pack()

def on_btn1_clicked():
    print("버튼1이 클릭되었습니다.")

btn1 = tk.Button(root, text="클릭해볼까요?", command=on_btn1_clicked)
btn1.pack()

root.mainloop()
```

# 위젯 종류
- 기본위젯(Widgets) 종류



- Label(라벨): 텍스트나이미지를화면에표시하는읽기전용위젯입니다.


- Button(버튼): 사용자가클릭했을때특정함수나명령을실행하는버튼입니다.


- Entry (엔트리): 한 줄의텍스트를입력받을수있는입력창입니다.


- Text(텍스트): 여러줄의텍스트를입력하거나출력할수있는편집기형태의위젯입니다.


- Checkbutton (체크버튼): 여러 개중0개이상을중복해서선택할수있는체크박스입니다.


- Radiobutton (라디오버튼): 여러개중단하나의옵션만선택할수있는라디오버튼입니다.


- Listbox (리스트박스): 여러줄의항목목록을보여주고선택하게하는목록상자입니다.


- Combobox(콤보박스): ttk 모듈에있는항목목록드롭다운선택창입니다.


- Scale (스케일): 슬라이더를마우스로끌어연속된숫자값을선택하는위젯입니다.


- Scrollbar (스크롤바): 텍스트나리스트박스, 캔버스등에연결해화면을스크롤하는위젯입니다.


- Canvas (캔버스): 선, 도형, 이미지등을자유롭게그리고배치할수있는도화지영역입니다.


- Frame(프레임): 다른위젯들을담는컨테이너(구조물) 역할의영역입니다.



- Toplevel (탑레벨): 메인 창외에새로생성되는독립된보조창(팝업창)입니다.



- Menu/ Menubutton (메뉴): 상단 메뉴바나드롭다운메뉴를만드는위젯입니다. [1
, 2
]

## Lab: 화씨 섭씨 

```python
from tkinter import *

def process():
    try:
        tf = float(e1.get())
        tc = (tf - 32.0) * 5.0 / 9.0
        e2.delete(0, END)
        e2.insert(0, f"{tc:.2f}")
    except ValueError:
        e2.delete(0, END)
        e2.insert(0, "숫자 입력")

window = Tk()
window.title("온도 변환기")

Label(window, text="화씨").grid(row=0, column=0, padx=5, pady=5)
Label(window, text="섭씨").grid(row=1, column=0, padx=5, pady=5)

e1 = Entry(window)
e2 = Entry(window)
e1.grid(row=0, column=1, padx=5, pady=5)
e2.grid(row=1, column=1, padx=5, pady=5)

Button(window, text="화씨->섭씨", command=process).grid(row=2, column=1, padx=5, pady=5)

window.mainloop()
```

## Lab:숫자추측게임

```python
from tkinter import *
import random

answer = random.randint(1, 100)

def guessing():
    try:
        guess = int(guessField.get())
        if guess > answer:
            msg = "높음!"
        elif guess < answer:
            msg = "낮음!"
        else:
            msg = "정답!"
        resultLabel["text"] = msg
    except ValueError:
        resultLabel["text"] = "숫자를 입력하세요!"
    guessField.delete(0, END)

def reset():
    global answer
    answer = random.randint(1, 100)
    resultLabel["text"] = "다시 한번 하세요!"
    guessField.delete(0, END)

window = Tk()
window.configure(bg="white")
window.title("숫자를 맞춰보세요!")
window.geometry("500x100")

titleLabel = Label(window, text="숫자 게임에 오신 것을 환영합니다!", bg="white")
titleLabel.pack(pady=5)

frame = Frame(window, bg="white")
frame.pack(pady=5)

guessField = Entry(frame, width=10)
guessField.pack(side="left", padx=5)

tryButton = Button(frame, text="시도", fg="green", bg="white", command=guessing)
tryButton.pack(side="left", padx=2)

resetButton = Button(frame, text="초기화", fg="red", bg="white", command=reset)
resetButton.pack(side="left", padx=2)

resultLabel = Label(frame, text="1부터 100 사이의 숫자를 입력하시오.", bg="white")
resultLabel.pack(side="left", padx=5)

window.mainloop()
```

## Lab: TIC-TAC-TOE 게임

```python
from tkinter import *

def checked(i):
    global player
    button = button_list[i]
    if button["text"] != " ":
        return
    button["text"] = player
    if player == "X":
        button["bg"] = "yellow"
        player = "O"
    else:
        button["bg"] = "lightgreen"
        player = "X"

window = Tk()
player = "X"
button_list = []

for i in range(9):
    b = Button(window, text=" ", width=6, height=3, command=lambda k=i: checked(k))
    b.grid(row=i//3, column=i%3)
    button_list.append(b)

window.mainloop()
```

## Lab: 계산기프로그램

```python
from tkinter import *

def click(key):
    if key == "=":
        try:
            result = eval(display.get())
            display.delete(0, END)
            display.insert(END, str(result))
        except:
            display.delete(0, END)
            display.insert(END, "Error")
    elif key == "C":
        display.delete(0, END)
    elif key != " ":
        display.insert(END, key)

window = Tk()
window.title("My Calculator")

display = Entry(window, width=33, bg="yellow")
display.grid(row=0, column=0, columnspan=5)

button_list = [
    '7', '8', '9', '/', 'C',
    '4', '5', '6', '*', ' ',
    '1', '2', '3', '-', ' ',
    '0', '.', '=', '+', ' '
]

row_index = 1
col_index = 0

for button_text in button_list:
    cmd = lambda x=button_text: click(x)
    Button(window, text=button_text, width=5, command=cmd).grid(row=row_index, column=col_index)
    col_index += 1
    if col_index > 4:
        row_index += 1
        col_index = 0

window.mainloop()
```

## Lab: 도형 그리기

```python
from tkinter import *

WIDTH = 600
HEIGHT = 200

def displayRect():
    canvas.create_rectangle(10, 10, WIDTH - 10, HEIGHT - 10)

def displayOval():
    canvas.create_oval(10, 10, WIDTH - 10, HEIGHT - 10, fill="yellow")

def displayArc():
    canvas.create_arc(10, 10, WIDTH - 10, HEIGHT - 10, start=0, extent=120, width=10, fill='blue')

def displayPolygon():
    canvas.create_polygon(10, 10, WIDTH - 10, HEIGHT - 10, 200, 90, 300, 160)

def displayLine():
    canvas.create_line(10, 10, WIDTH - 10, HEIGHT - 10, fill='green')

def clearCanvas():
    canvas.delete(ALL)

window = Tk()
canvas = Canvas(window, width=WIDTH, height=HEIGHT, bg='white')
canvas.pack()

frame = Frame(window)
frame.pack()

Button(frame, text="Line", command=displayLine).grid(row=1, column=1)
Button(frame, text="Rectangle", command=displayRect).grid(row=1, column=2)
Button(frame, text="Oval", command=displayOval).grid(row=1, column=3)
Button(frame, text="Polygon", command=displayPolygon).grid(row=1, column=4)
Button(frame, text="Arc", command=displayArc).grid(row=1, column=5)
Button(frame, text="Clear", command=clearCanvas).grid(row=1, column=6)

window.mainloop()
```

## 마우스이벤트처리

```python
from tkinter import*
window = Tk()
def key(event):
    print( repr(event.char), "가눌렸습니다. ")

def callback(event):
    frame.focus_set()
    print(event.x, event.y, "에서마우스이벤트발생")
    
frame = Frame(window, width=100, height=100)
frame.bind("<Key>", key)
frame.bind("<Button-1>", callback)
frame.pack()
window.mainloop()
```

## Lab: 그림판 프로그램만들기

```python
from tkinter import *
from tkinter.colorchooser import askcolor

mode = "pen"
mycolor = "black"
old_x = None
old_y = None

def paint(event):
    global old_x, old_y
    fill_color = "white" if mode == "erase" else mycolor
    if old_x is not None and old_y is not None:
        canvas.create_line(old_x, old_y, event.x, event.y, capstyle=ROUND, width=var.get(), fill=fill_color)
    old_x = event.x
    old_y = event.y

def reset(event):
    global old_x, old_y
    old_x, old_y = None, None

def use_pen():
    global mode
    mode = "pen"

def use_brush():
    global mode
    mode = "brush"

def choose_color():
    global mycolor
    color = askcolor(color=mycolor)[1]
    if color:
        mycolor = color

def use_eraser():
    global mode
    mode = "erase"

def clear_all():
    canvas.delete("all")

window = Tk()
window.title("그림판")

penButton = Button(window, text='펜', command=use_pen)
penButton.grid(row=0, column=0, sticky=W+E)

brushButton = Button(window, text='브러쉬', command=use_brush)
brushButton.grid(row=0, column=1, sticky=W+E)

colorButton = Button(window, text='색상선택', command=choose_color)
colorButton.grid(row=0, column=2, sticky=W+E)

eraseButton = Button(window, text='지우개', command=use_eraser)
eraseButton.grid(row=0, column=3, sticky=W+E)

clearButton = Button(window, text='모두삭제', command=clear_all)
clearButton.grid(row=0, column=4, sticky=W+E)

var = DoubleVar(value=5)
scale = Scale(window, variable=var, orient=VERTICAL, from_=1, to=50)
scale.grid(row=1, column=5, sticky=N+S)

canvas = Canvas(window, bg='white', width=600, height=400)
canvas.grid(row=1, column=0, columnspan=5)

canvas.bind('<B1-Motion>', paint)
canvas.bind('<ButtonRelease-1>', reset)

window.mainloop()
```

## Lab: 공 애니메이션

```python
from tkinter import *
import time
import random

window = Tk()
canvas = Canvas(window, width=600, height=400)
canvas.pack()
window.update()

class Ball:
    def __init__(self, color, size):
        self.id = canvas.create_oval(0, 0, size, size, fill=color)
        self.dx = random.randint(1, 10)
        self.dy = random.randint(1, 10)

    def move(self):
        canvas.move(self.id, self.dx, self.dy)
        x0, y0, x1, y1 = canvas.coords(self.id)
        if y1 > canvas.winfo_height() or y0 < 0:
            self.dy = -self.dy
        if x1 > canvas.winfo_width() or x0 < 0:
            self.dx = -self.dx

ball1 = Ball("blue", 60)
ball2 = Ball("green", 100)
ball3 = Ball("orange", 80)

try:
    while True:
        ball1.move()
        ball2.move()
        ball3.move()
        window.update()
        time.sleep(0.05)
except TclError:
    pass
```
