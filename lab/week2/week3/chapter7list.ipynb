### 리스트 연산들
- append() : 공백 리스트를 먼저 생성해놓고 나중에 추가할 때 유용  
- insert() : 지정된 위치에 요소를 추가할 때 유용  
- index() : 특정 항목이 리스트의 어디에 있는지 알 때 유용  
- pop() : 항목이 지정된 위치를 알고 있을 때 유용  
- remove() : 항목의 값만 알고 있을 때 유용  
- random.choice :  리스트에서 랜덤으로 선택할 때 유용
