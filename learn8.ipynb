# counter 클래스 만들기 
- 데이터: 현재 숫자
- 동작: 증가하기, 현재 값 읽어오기

```python
class Counter:
    def __init__(self, initial_count=0):  # 생성함수
        self.count = initial_count
    
    def increment(self, amount=1):  # 증가함수
        self.count += amount
    
    def get_count(self):  # 카운트값 반환함수
        return self.count


def test_counter():
    myc = Counter()
    print(myc.get_count())
    
    myc.increment()
    print(myc.get_count())
    
    myc.increment()
    print(myc.get_count())
    
    myc.increment()
    print(myc.get_count())
    
    for i in range(10):
        myc.increment()
    print(myc.get_count())

    yourc = Counter(5)
    print(yourc.get_count())
    
    yourc.increment()
    print(yourc.get_count())


if __name__ == "__main__":
    test_counter()

```

# TV 클래스 추가 
- 데이터: 현재 채널번호, 볼륨량, on 여부
- 동작: 켜기,끄기,채널변경,볼륨변경,현재 TV 객체 내용 출력

-클랙스명: MyTelevison



```python
class MyTelevision:
    def __init__(self, channel=1, volume=10, is_on=False):  # 생성함수
        self.channel = channel
        self.volume = volume
        self.is_on = is_on

    def set_turn_on(self):  # 전원 켜기
        self.is_on = True

    def set_turn_off(self):  # 전원 끄기
        self.is_on = False

    def change_channel(self, channel):  # 채널 설정함수
        self.channel = channel

    def change_volume(self, volume):  # 볼륨 설정함수
        self.volume = volume

    def display_status(self):  # 상태 출력함수
        if self.is_on:
            print(f"Channel: {self.channel}, Volume: {self.volume}, Is_on: {self.is_on}")
        else:
            print("TV is OFF")


def test_television():
    my_tv = MyTelevision(channel=11)

    my_tv.set_turn_on()
    my_tv.display_status()

    my_tv.set_turn_off()
    my_tv.display_status()


if __name__ == "__main__":
    test_television()

```

# TV 리모콘 클래스

- 데이터는 없고 동작만 있음
- 연관된 TV를 켜거나 꺼라는 동작만 지원
- TV 상태 표시 추가

```python
class tvRemocon:
    def __init__(self, tv: MyTelevision):  # 생성함수
        self.tv = tv

    def press_on(self):  # 전원 켜기
        self.tv.set_turn_on()

    def press_off(self):  # 전원 끄기
        self.tv.set_turn_off()

    def press_display(self):  # TV 상태 확인
        self.tv.display_status()


def test_remocon():
    tv1 = MyTelevision()
    rc1 = tvRemocon(tv1)

    rc1.press_on()
    rc1.press_display()

    rc1.press_off()
    rc1.press_display()


if __name__ == "__main__":
    test_remocon()

```
