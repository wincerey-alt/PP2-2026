# 12장 상속

- 상속에 대한 이론과 실제 학습 

```python
class Car:
    def __init__(self, maker, model, color, price):
        self.maker = maker  
        self.model = model
        self.color = color
        self.price = price

    def set_maker(self, maker):
        self.maker = maker

    def get_maker(self):
        return self.maker

    def get_desc(self):
        return str(self)

    def __str__(self):
        return f"Maker: {self.maker}, Model: {self.model}, Color: {self.color}, Price: {self.price}"


class ElectricCar(Car):
    def __init__(self, maker, model, color, price, bat_size):
        super().__init__(maker, model, color, price)
        self.battery_size = bat_size

    def get_battery_size(self):
        return self.battery_size

    def set_battery_size(self, bat_size):
        self.battery_size = bat_size

    def __str__(self):
        return f"{super().__str__()}, Battery: {self.battery_size}"


class Truck(Car):
    def __init__(self, maker, model, color, price, payload):
        super().__init__(maker, model, color, price)
        self.payload = payload

    def __str__(self):
        return f"{super().__str__()}, PayLoad: {self.payload}"


# 실행 테스트
car1 = Car("Kia", "Ray", "Blue", 2000)
print(car1.get_desc())

ev1 = ElectricCar("Tesla", "Model 5", "Red", 8000, "75kWh")
print(ev1.get_desc())

truck1 = Truck("Hyundai", "Porter", "White", 4000, "1 ton")
print(truck1.get_desc())

```
