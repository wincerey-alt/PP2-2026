# 객체지향의3가지핵심요소
- 캡슐화: 데이터와연산을하나의덩어리로묶어서다룸

- 정보은닉을위한가시성: public (공개), protected (보호), private(숨김)

- 다른클래스에서이클래스의데이터와연산을접근하거나접근불가로만듬

- Public: 모든 클래스에서접근가능, protected: 자식만접근가능,  -private : 접근 불가

- 상속: 부모클래스를자식클래스가그대로물려받아추가적인코드를작성할수있음

- 다만, 부모의private 한 것은상속은받으나접근불가능
- Protected는 외부 접근불가능하나자식이물려받아사용가능
- 다형성: 함수이름이같더라도문맥(혹은클래스)에따라다른함수가수행됨

-  A.do(), B.do() 처럼 do()함수 이름은 같으나 클래스가다르면다른함수
-  B가A를상속하고재정의할경우동적바인딩이가능


```python
class Car:
    def __init__(self, make, model, color, price):
        # 4칸 들여쓰기로 클래스 및 생성자 블록 맞추기
        self.make = make
        self.model = model
        self.color = color
        self.price = price

    # 메서드들도 Car 클래스 내부에 포함되도록 들여쓰기
    def setMake(self, make):
        self.make = make

    def getMake(self):
        # 값을 반환(return)하도록 수정
        return self.make

    def getDesc(self):
        # f-string을 사용해 간결하고 명확하게 표현
        return f"차량=({self.make},{self.model},{self.color},{self.price})"


# 실행 테스트
my_car = Car("현대", "아반떼", "흰색", 2000)
print(my_car.getDesc())  # 차량=(현대,아반떼,흰색,2000)

my_car.setMake("기아")
print(my_car.getMake())  # 기아
print(my_car.getDesc())  # 차량=(기아,아반떼,흰색,2000)
```

```python
class Animal:
    def __init__(self, age=0):
        self.age = age

    def eat(self):
        print("동물이 먹고 있습니다.")


class Dog(Animal):
    def __init__(self, age=0, name=""):
        # 부모 클래스의 생성자를 명시적으로 호출하여 self.age를 초기화
        super().__init__(age)
        self.name = name


# 정상 동작 테스트
d = Dog(age=3, name="바둑이")
print(d.age)   # 3
print(d.name)  # 바둑이
d.eat()        # 동물이 먹고 있습니다.
```

```python


class Bank():
    def getInterestRate(self):
        return 0.0
    
class BadBank(Bank):
    def getInterestRate(self):
        return 10.0;

class NormalBank(Bank):
    def getInterestRate(self):
        return 5.0;

class GoodBank(Bank):
    def getInterestRate(self):
        return 3.0;

b1 = BadBank()
b2 = NormalBank()
b3 = GoodBank()
print("BadBank의이자율: "+ str(b1.getInterestRate()))
print("NormalBank의이자율: "+ str(b2.getInterestRate()))
print("GoodBank의이자율: "+ str(b3.getInterestRate()))
```

# 문제: 일반적인사람을나타내는Person 클래스를정의한다. Person 클래스를상속받아서학생을나타내는
클래스Student와선생님을나타내는클래스Teacher를정의한다. 

```python
class Person:
    def __init__(self, name, number):
        self.name = name
        self.number = number


class Student(Person):
    UNDERGRADUATE = 0
    POSTGRADUATE = 1

    def __init__(self, name, number, studentType):
        super().__init__(name, number)
        self.studentType = studentType
        self.gpa = 0
        self.classes = []

    def enrollCourse(self, course):
        self.classes.append(course)

    def __str__(self):
        return f"\n이름={self.name}\n주민번호={self.number}\n수강과목={self.classes}\n평점={self.gpa}"


class Teacher(Person):
    def __init__(self, name, number):
        super().__init__(name, number)
        self.courses = []
        self.salary = 3000000

    def assignTeaching(self, course):
        self.courses.append(course)

    def __str__(self):
        return f"\n이름={self.name}\n주민번호={self.number}\n강의과목={self.courses}\n월급={self.salary}"


hong = Student("홍길동", "12345678", Student.UNDERGRADUATE)
hong.enrollCourse("자료구조")
print(hong)

kim = Teacher("김철수", "123456790")
kim.assignTeaching("Python")
print(kim)
```

# 문제: 은행계좌를나타내는클래스BankAccount 클래스를정의한다. 

```python
class BankAccount:
    def __init__(self, name, number, balance):
        self.balance = balance
        self.name = name
        self.number = number

    def withdraw(self, amount):
        self.balance -= amount
        return self.balance

    def deposit(self, amount):
        self.balance += amount
        return self.balance


class SavingsAccount(BankAccount):
    def __init__(self, name, number, balance, interest_rate):
        super().__init__(name, number, balance)
        self.interest_rate = interest_rate

    def set_interest_rate(self, interest_rate):
        self.interest_rate = interest_rate

    def get_interest_rate(self):
        return self.interest_rate

    def add_interest(self):
        self.balance += self.balance * self.interest_rate


class CheckingAccount(BankAccount):
    def __init__(self, name, number, balance):
        super().__init__(name, number, balance)
        self.withdraw_charge = 10000

    def withdraw(self, amount):
        return super().withdraw(amount + self.withdraw_charge)


a1 = SavingsAccount("홍길동", 123456, 10000, 0.05)
a1.add_interest()
print("저축예금의잔액=", a1.balance)

a2 = CheckingAccount("김철수", 123457, 2000000)
a2.withdraw(100000)
print("당좌예금의잔액=", a2.balance)
```
